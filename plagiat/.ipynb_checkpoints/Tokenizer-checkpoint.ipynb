```python
import re
import os
```

```python
def tokenize(text):
    token_pattern = r"""
        (?:\b\d+(?:\.\d+)?\b)        
        | \w+(?:[-'’]\w+)?            
        | [.,!?;:]    
    """
    text = text.lower()
    pattern = re.compile(token_pattern, re.VERBOSE)
    
    tokens = pattern.findall(text)
    return tokens
```

```python
test_cases = [
    "Hello, world! It's 2025. Don't forget to check $10.50.",
    "Visit https://example.com for more info. #NLP @OpenAI",
    "The U.S.A. won 3.14 times in a row.",
    "Prices: $1,000.00; Dates: 01/01/2025.",
    "Hello\nworld! This is a multi-line test."
]

for case in test_cases:
    print(f"Input: {case}")
    print(f"Output: {tokenize(case)}\n")
```

```python
def read_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        content = f.read()
    return content
```

```python
from typing import List, Dict
import numpy as np
```

```python
from collections import Counter

class CountVectorFeaturizer:
    def __init__(self):
        self.vocabulary = {}  # Maps tokens to indices

    def fit(self, documents: List[str]) -> None:
        """
        Build the vocabulary from the corpus.

        Args:
            documents (List[str]): List of text documents.
        """
        # Create a Counter for token frequencies
        token_counter = Counter()
        
        # Tokenize and count token frequencies
        for doc in documents:
            tokens = tokenize(doc)  # Implement your tokenizer function
            token_counter.update(tokens)
        
        # Map each token to a unique index based on frequency
        self.vocabulary = {token: idx for idx, (token, _) in enumerate(token_counter.most_common())}

    def transform(self, documents: List[str]) -> np.ndarray:
        """
        Transform documents into count vectors.

        Args:
            documents (List[str]): List of text documents.

        Returns:
            np.ndarray: Sparse matrix of shape (num_documents, vocabulary_size).
        """
        # Initialize a zero matrix for count vectors
        num_docs = len(documents)
        vocab_size = len(self.vocabulary)
        matrix = np.zeros((num_docs, vocab_size), dtype=int)

        # Populate the matrix with token counts
        for i, doc in enumerate(documents):
            tokens = tokenize(doc)
            for token in tokens:
                if token in self.vocabulary:
                    matrix[i, self.vocabulary[token]] += 1

        return matrix

    def fit_transform(self, documents: List[str]) -> np.ndarray:
        """
        Fit the vocabulary and transform documents in one step.

        Args:
            documents (List[str]): List of text documents.

        Returns:
            np.ndarray: Sparse matrix of shape (num_documents, vocabulary_size).
        """
        self.fit(documents)
        return self.transform(documents)
```

```python
folder_path = "../data-plagiarism"
documents = []

featurizer = CountVectorFeaturizer()
for file_name in os.listdir(folder_path):
    if file_name.endswith(".txt"): 
        file_path = os.path.join(folder_path, file_name)
        content = read_file(file_path)
        documents.append(content)
        #print(f"Tokens for {file_name}:")
        #tokens = tokenize(content)
        #print(tokens)
        #print("\n") 
count_matrix = featurizer.fit_transform(documents)
print("Vocabulary:", featurizer.vocabulary)
print("Count Matrix:\n", count_matrix)
print("\n") 
```
